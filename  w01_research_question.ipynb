1) My lane (or freestyle) and why
My Lane: Content Refresh Queue (Predicting Declining Pages)
Why: This lane aligns perfectly with the provided starter dataset. We have historical data on pages, and predicting traffic decay allows us to build a direct, prioritized queue of pages that need updating.


2) The question: decision, action, cost of a wrong call

The Decision: Which existing pages on the website should we prioritize for a content refresh this month?

The Action: Assigning a human writer/editor to update or improve the flagged page.

Cost of a wrong call: False Positives waste expensive writer hours. False Negatives mean we lose organic traffic to competitors.

Why ML helps: Simple hand-written rules only achieve ~24% precision. ML can roughly triple that precision.
