1) My lane as an ML task (type)
Lane: Content Refresh Queue
Task Type: Binary Classification. We are classifying whether a webpage is "decaying" (needs a refresh) or "stable/growing" (leave it alone).

2) Target or proxy
Target: The exact target variable is whether the page's organic traffic trend is heading downwards over a sustained period (e.g., a label of 1 for declining, 0 for stable/growing).

3) Success metric
Metric: Precision and Recall (specifically, the F1-Score).

We need high Precision so we don't assign writers to update pages that are actually doing fine (wasting money).

We need good Recall so we don't completely miss pages that are bleeding traffic and urgently need help.

5) Why ML beats a fixed rule here
A fixed rule (like "flag if traffic drops 10% in 30 days") is too brittle. It flags pages that drop due to normal seasonal trends (like holiday pages in January) and misses pages experiencing a slow, creeping decay. ML can look at multiple features at once—seasonality, backlink profiles, content age, and search volume—to separate normal fluctuations from genuine content decay.
